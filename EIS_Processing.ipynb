```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy import signal, stats, optimize

#change file name
df = pd.read_excel(r"C:\Users\mtpv1\Downloads\Test.xlsx")

average_impedance = (
    df.groupby(["Surface_Finish", "Area_um2", "Frequency_Hz"])["Impedance_Ohm"]
    .mean()
    .reset_index()
)

for finish, finish_df in average_impedance.groupby("Surface_Finish"):
    plt.figure()

    for area, area_df in finish_df.groupby("Area_um2"):
        plt.plot(
            area_df["Frequency_Hz"],
            area_df["Impedance_Ohm"],
            marker = 'o',
            label = f"{area} µm²"
        )

    plt.xscale("log")
    plt.yscale("log")
    plt.xlabel("Frequency (Hz)")
    plt.ylabel("Average Impedance (Ohms)")
    plt.title(f"Average Impedance vs. Frequency: {finish}")
    plt.legend(title="Electrode area")
    plt.grid(True, which="both")
    plt.show()

average_phase_shift = (
    df.groupby(["Surface_Finish", "Area_um2", "Frequency_Hz"])["Phase_Shift"]
    .mean()
    .reset_index()
)

for finish, finish_df in average_phase_shift.groupby("Surface_Finish"):
    plt.figure()

    for area, area_df in finish_df.groupby("Area_um2"):
        plt.plot(
            area_df["Frequency_Hz"],
            area_df["Phase_Shift"],
            marker = 'o',
            label = f"{area} µm²"
        )

    plt.xscale("log")
    plt.xlabel("Frequency (Hz)")
    plt.ylabel("Phase Shift (Rad)")
    plt.title(f"Average Phase Shift vs. Frequency: {finish}")
    plt.legend(title="Electrode area")
    plt.grid(True, which="both")
    plt.show()


std_impedance = (
    df.groupby(["Surface_Finish", "Area_um2", "Frequency_Hz"])["Impedance_Ohm"]
    .std()
    .reset_index()
)

for finish, finish_df in std_impedance.groupby("Surface_Finish"):
    plt.figure()

    for area, area_df in finish_df.groupby("Area_um2"):
        plt.plot(
            area_df["Frequency_Hz"],
            area_df["Impedance_Ohm"],
            marker = 'o',
            label = f"{area} µm²"
        )

    plt.xscale("log")
    plt.xlabel("Frequency (Hz)")
    plt.ylabel("Standard Deviation (Ohms)")
    plt.title(f"Standard Deviation of Impedance vs. Frequency: {finish}")
    plt.legend(title="Electrode area")
    plt.grid(True, which="both")
    plt.show()
```
