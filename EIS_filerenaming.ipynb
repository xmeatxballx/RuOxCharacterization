```python
from pathlib import Path

# ============================================================
# USER SETTING
# ============================================================

folder = Path(r"D:\EIS-Data\4_1_Au_300")

frequencies = [
    1,
    10,
    100,
    1000,
    10000
]

# ============================================================
# GET FOLDER NAME
# ============================================================

# Example:
# C:\...\4_1_Al_100
# becomes:
# 4_1_Al_100

folder_name = folder.name


# ============================================================
# FIND RAW CSV FILES
# ============================================================

files = list(folder.glob("*.csv"))


# Extract the ending number:
# SpikeIoT_12_Detail261.csv -> 261
def ending_number(file):
    stem = file.stem

    digits = ""

    for character in reversed(stem):
        if character.isdigit():
            digits = character + digits
        else:
            break

    return int(digits)


# Sort 261, 262, 263, ...
files = sorted(files, key=ending_number)


# ============================================================
# GENERATE NEW NAMES
# ============================================================

new_names = []

for frequency in frequencies:

    # Channel 1 = generator / electrode voltage
    new_names.append(
        f"{folder_name}_{frequency}_1.csv"
    )

    # Channel 2 = sense resistor voltage
    new_names.append(
        f"{folder_name}_{frequency}_2.csv"
    )


# ============================================================
# SAFETY CHECK
# ============================================================

if len(files) != len(new_names):
    raise ValueError(
        f"Found {len(files)} CSV files, "
        f"but expected {len(new_names)}."
    )


# ============================================================
# PREVIEW
# ============================================================

print("Planned renaming:\n")

for old_file, new_name in zip(files, new_names):
    print(
        f"{old_file.name}  -->  {new_name}"
    )


# ============================================================
# CONFIRM
# ============================================================

answer = input(
    "\nProceed with renaming? (y/n): "
)

if answer.lower() == "y":

    for old_file, new_name in zip(files, new_names):

        new_file = folder / new_name

        old_file.rename(new_file)

    print("\nRenaming complete.")

else:
    print("\nNothing was renamed.")
```
